```python
from pathlib import Path
import json, py_compile
PROJECT = Path('projects/energy_demand_forecasting')
if not PROJECT.exists(): PROJECT = Path.cwd()
print((PROJECT / 'README.md').read_text(encoding='utf-8')[:2500])
```

```python
for path in sorted(PROJECT.rglob('*.py')):
    py_compile.compile(str(path), doraise=True)
    print('✓', path.relative_to(PROJECT))
```

```python
metrics = json.loads((PROJECT / 'results' / 'verified_metrics.json').read_text(encoding='utf-8'))
print(json.dumps(metrics, indent=2)[:5000])
```
