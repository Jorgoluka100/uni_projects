# Parkinson's Progression Modelling

**Hiring signal:** Data Science · Validation Design

**Dataset:** UCI Parkinson's Telemonitoring. `run.py` accepts the official UCI source or a local `parkinsons_updrs.data` copy, and evaluation is grouped by subject to avoid repeated-measure leakage. This is an educational modelling project, not a clinical decision system.


```python
from pathlib import Path
import json, sys, subprocess
import pandas as pd
PROJECT_REL = Path('projects/parkinsons_progression')
ROOT = Path.cwd()
if not (ROOT / PROJECT_REL).exists() and (ROOT / 'uni_projects' / PROJECT_REL).exists(): ROOT = ROOT / 'uni_projects'
PROJECT = ROOT / PROJECT_REL
assert PROJECT.exists(), 'Open from the cloned uni_projects repository.'
print('Project:', PROJECT.resolve())
```

## 1. Dataset and leakage-safe validation


```python
readme = (PROJECT / 'README.md').read_text(encoding='utf-8')
print(readme[:8500])
```

## 2. Implementation and grouped evaluation evidence


```python
for pattern in ('run.py','src/*.py','tests/*.py'):
    for path in sorted(PROJECT.glob(pattern)): print(f'{path.relative_to(PROJECT)}  ({path.stat().st_size:,} bytes)')
verified = ROOT / 'verified/parkinsons_grouped'
for name in ('evaluation.json','verification.json'):
    path = verified / name
    if path.exists():
        print(f'\n--- {name} ---')
        print(json.dumps(json.loads(path.read_text(encoding='utf-8')), indent=2)[:12000])
for path in sorted(verified.glob('*.csv'))[:3]: display(pd.read_csv(path).head(10))
```

## 3. Reproduce


```python
RUN_PIPELINE = False
if RUN_PIPELINE: subprocess.run([sys.executable, 'run.py'], cwd=PROJECT, check=True)
else: print('Rebuild: cd projects/parkinsons_progression && python run.py')
```
