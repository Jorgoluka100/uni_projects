# Parkinson's Progression — Project Notebook

Notebook companion to the leakage-aware Python project. The `.py` files remain the source of truth; this notebook makes the workflow easy to inspect in Jupyter/Colab without duplicating the implementation.

````python
from pathlib import Path
import json

PROJECT = Path('projects/parkinsons_progression')
if not PROJECT.exists():
    PROJECT = Path.cwd()
print('Project:', PROJECT.resolve())
print('Python files:')
for path in sorted(PROJECT.rglob('*.py')):
    print(' -', path)
````

````python
readme = PROJECT / 'README.md'
if readme.exists():
    print(readme.read_text(encoding='utf-8')[:12000])
````

````python
print('Evaluation/result files:')
for path in sorted(PROJECT.rglob('*.json')):
    print('\n---', path, '---')
    try:
        print(json.dumps(json.loads(path.read_text(encoding='utf-8')), indent=2)[:8000])
    except Exception:
        print(path.read_text(encoding='utf-8')[:2000])
````
