# Business Intelligence Showcase — Project Notebook

Companion notebook for the Power BI and Tableau portfolio projects. The dashboards reuse the verified Olist SQL/DuckDB warehouse instead of creating a second set of business rules. This notebook is for inspecting data lineage and dashboard assets; the actual BI deliverables live in `power_bi/` and `tableau/`.

```python
from pathlib import Path
import json
PROJECT = Path('projects/business_intelligence_showcase')
if not PROJECT.exists(): PROJECT = Path.cwd()
print('Project:', PROJECT.resolve())
for path in sorted(PROJECT.rglob('*')):
    if path.is_file() and path.suffix.lower() in {'.py','.pbip','.pbir','.pbism','.tmdl','.twb','.svg','.csv','.json'}:
        print(' -', path.relative_to(PROJECT))
```

```python
data = PROJECT / 'data'
manifest = data / 'manifest.json'
if manifest.exists():
    print(json.dumps(json.loads(manifest.read_text(encoding='utf-8')), indent=2))
else:
    print('Run: python projects/business_intelligence_showcase/build_exports.py')
```

```python
for folder in ('power_bi', 'tableau'):
    readme = PROJECT / folder / 'README.md'
    if readme.exists():
        print(f'\n--- {folder} ---')
        print(readme.read_text(encoding='utf-8')[:8000])
```
