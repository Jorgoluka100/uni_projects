# 02 — NumPy for Machine Learning

The point of NumPy is not memorising functions. It is understanding shapes, vectorised numerical operations and the matrix calculations underneath many ML models.

```python
import numpy as np

X = np.array([
    [1.0, 20.0, 0.2],
    [1.0, 35.0, 0.7],
    [1.0, 50.0, 1.1],
    [1.0, 65.0, 1.5],
])
weights = np.array([-2.0, 0.04, 1.2])

print('X shape:', X.shape)
print('weights shape:', weights.shape)
scores = X @ weights
print('matrix multiplication:', scores)
```

```python
# Broadcasting: standardise every column without a Python loop
features = X[:, 1:]
mean = features.mean(axis=0)
std = features.std(axis=0)
standardised = (features - mean) / std
print('means after scaling:', standardised.mean(axis=0).round(6))
print('std after scaling:', standardised.std(axis=0).round(6))
```

```python
# Vectorised sigmoid: turn arbitrary scores into values between 0 and 1
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

probabilities = sigmoid(scores)
predictions = (probabilities >= 0.5).astype(int)
print('probabilities:', probabilities.round(3))
print('predictions:', predictions)
```

## Why this matters

A dense neural-network layer is fundamentally a matrix multiplication plus a bias followed by an activation. Understanding `shape`, `axis`, broadcasting and `@` makes PyTorch/TensorFlow code much easier to reason about instead of treating the framework as magic.
